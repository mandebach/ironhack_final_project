```python
# Load the comprehensive dataset from Notebook 2
energy_minute_df = pd.read_parquet("clean_master_with_actuals.parquet")
```

## 3.2 Dynamic Energy Pricing for the Netherlands

To pull live or historical dynamic energy prices for the Netherlands into your pipeline, the gold standard source used by data engineers across Europe is the ENTSO-E Transparency Platform API (the official clearinghouse for European wholesale electricity data).

<i>Source</i>: <a href="transparency.entsoe.eu">ENTSO-E Transparency Platform</a>

<i>Fields</i>: spot_price_eur_kwh, is_negative_pricing

<b>Negative Pricing</b>: Wholesale spot prices drop below zero when renewable generation (solar/wind) significantly exceeds grid demand. During these intervals, grid injection incurs a cost rather than earning revenue.

```python
# Load the .env file with the keys
load_dotenv()
client = Client(api_key=os.environ.get("ENTSOE_API_KEY"))
```

```python
# Define the full-year timeframe
start_date = pd.Timestamp("2025-10-01", tz="Europe/Brussels")
end_date = pd.Timestamp("2026-11-01", tz="Europe/Brussels") # End is 1 November because freq='Ms' (month start)

# Generate monthly start dates to chunk the requests safely
month_ranges = pd.date_range(start=start_date, end=end_date, freq='MS')

all_market_dfs = []

# Loop through month by month
for i in range(len(month_ranges) - 1):
    chunk_start = month_ranges[i]
    chunk_end = month_ranges[i+1]
    
    print(f"Fetching market prices for {chunk_start.strftime('%Y-%m')}...")
    
    try:
        # Pull the month chunk
        df_chunk = client.prices.day_ahead(chunk_start, chunk_end, country="NL")
        
        # Clean and convert each chunk immediately
        df_temp = df_chunk.reset_index()
        time_col = 'timestamp' if 'timestamp' in df_temp.columns else 'index'
        price_col = 'value' if 'value' in df_temp.columns else df_temp.columns[1]
        
        df_temp = df_temp[[time_col, price_col]].copy()
        df_temp.columns = ["timestamp", "spot_price_eur_kwh"]
        
        all_market_dfs.append(df_temp)
    except Exception as e:
        print(f"Failed to fetch {chunk_start.strftime('%Y-%m')}: {e}")

# Concatenate all months into one master dataframe
df_dynamic_market = pd.concat(all_market_dfs, ignore_index=True)

# Final transformations (Unit conversion & negative pricing flag)
df_dynamic_market["spot_price_eur_kwh"] = df_dynamic_market["spot_price_eur_kwh"] / 1000.0
df_dynamic_market["is_negative_pricing"] = df_dynamic_market["spot_price_eur_kwh"] < 0

print("\n--- Full Year Loaded Successfully! ---")
print(f"Total hourly/15-min price rows: {len(df_dynamic_market)}")
print(df_dynamic_market.head())
```

```python
# Clean and ensure correct datetime formatting & sorting
energy_minute_df['utc_time'] = pd.to_datetime(energy_minute_df['utc_time'])
df_dynamic_market['timestamp'] = pd.to_datetime(df_dynamic_market['timestamp'])
```

## Merging the energy_minute_df with df_dynamic_market
Since the smart meter data is recorded every minute, and the market prices update every 15 minutes, the best way to combine them without losing resolution is a backward merge / asof merge or reindexing so that every 1-minute row inherits the active 15-minute market price of its current window.

```python
# Ensure UTC alignment (localize/convert safely)
if energy_minute_df['utc_time'].dt.tz is None:
    energy_minute_df['utc_time'] = energy_minute_df['utc_time'].dt.tz_localize('UTC')
else:
    energy_minute_df['utc_time'] = energy_minute_df['utc_time'].dt.tz_convert('UTC')

if df_dynamic_market['timestamp'].dt.tz is None:
    df_dynamic_market['timestamp'] = df_dynamic_market['timestamp'].dt.tz_localize('UTC')
else:
    df_dynamic_market['timestamp'] = df_dynamic_market['timestamp'].dt.tz_convert('UTC')

# CRITICAL: Sort both ascendingly so 'backward' lookups look forward in real time
energy_minute_df = energy_minute_df.sort_values('utc_time').reset_index(drop=True)
df_dynamic_market = df_dynamic_market.sort_values('timestamp').reset_index(drop=True)

# Perform the asof merge
merged_energy_df = pd.merge_asof(
    energy_minute_df,
    df_dynamic_market,
    left_on='utc_time',
    right_on='timestamp',
    direction='backward'
)

# Quick verification print to check matching dates
print(merged_energy_df[['utc_time', 'timestamp', 'spot_price_eur_kwh']].head(10))
```

```python
# Drop the redundant market timestamp column
merged_energy_df = merged_energy_df.drop(columns=['timestamp'])

print(merged_energy_df.columns.tolist())
```

## Dynamic Cost & Revenue Calculation

This simulation shows what the energy bills would have looked like under a dynamic contract (like ANWB Energie or Tibber). Step 1 is to
calculate the minute-level dynamic consumption costs and solar revenues.

For a dynamic energy contract in the Netherlands, your total cost per minute is typically:

Consumption Cost: Energy consumed ($\text{kWh}$) $\times$ (Spot Price + Supplier Markup + Government Energy Tax & VAT).

Injection Revenue: Energy injected back into the grid ($\text{kWh}$) $\times$ Spot Price (though subject to terugleverkosten).

```python
# Create a dedicated working copy for the dynamic simulation
df_dynamic = merged_energy_df.copy()

# Define the dynamic cost and revenue function
def calculate_dynamic_costs(row):
    # Convert minute kW to kWh
    kwh_consumed = row.get('power_consumed_kw', 0.0) / 60.0
    kwh_produced = row.get('power_produced_kw', 0.0) / 60.0
    
    spot_price = row.get('spot_price_eur_kwh', 0.0)
    
    # Typical dynamic supplier markup / delivery fee (e.g., ~€0.025 / kWh) + government energy tax on consumption (~€0.13/kWh)
    consumption_tax_and_markup = 0.155  
    solar_markup_deduction = 0.025 # Supplier fee deducted from your feed-in price
    
    dynamic_consumption_cost = kwh_consumed * (spot_price + consumption_tax_and_markup)
    
    # Dynamic solar revenue (if spot price is negative, you pay to export, minus the markup)
    effective_solar_rate = spot_price - solar_markup_deduction
    dynamic_solar_revenue = kwh_produced * effective_solar_rate
    
    return pd.Series([dynamic_consumption_cost, dynamic_solar_revenue])

# Apply the dynamic calculations across the working copy
df_dynamic[['dynamic_consumption_cost_eur', 'dynamic_solar_revenue_eur']] = df_dynamic.apply(calculate_dynamic_costs, axis=1)

print("Dynamic pricing simulation calculated successfully on df_dynamic!")
print(df_dynamic[['utc_time', 'spot_price_eur_kwh', 'dynamic_consumption_cost_eur', 'dynamic_solar_revenue_eur']].head())
```

## Monthly Totals

```python
# Use the already-loaded energy_minute_df to build monthly financials
energy_minute_df['year_month'] = energy_minute_df['utc_time'].dt.tz_localize(None).dt.to_period('M')

monthly_financials = energy_minute_df.groupby('year_month').agg({
    'energy_consumed_kwh': lambda x: (energy_minute_df.loc[x.index, 'power_consumed_kw'] / 60.0).sum(),
    'energy_produced_kwh': lambda x: (energy_minute_df.loc[x.index, 'power_produced_kw'] / 60.0).sum(),
    'actual_consumption_cost_eur': 'sum',
    'actual_solar_revenue_eur': 'sum'
}).reset_index()

monthly_financials['final_net_cost_eur'] = (
    monthly_financials['actual_consumption_cost_eur'] - 
    monthly_financials['actual_solar_revenue_eur']
)

# Aggregate dynamic costs and revenues by month from your simulation sandbox
df_dynamic['year_month'] = df_dynamic['utc_time'].dt.tz_localize(None).dt.to_period('M')

monthly_dynamic = df_dynamic.groupby('year_month').agg({
    'dynamic_consumption_cost_eur': 'sum',
    'dynamic_solar_revenue_eur': 'sum'
}).reset_index()

# Merge for a direct side-by-side comparison
comparison_df = pd.merge(monthly_financials[['year_month', 'final_net_cost_eur']], monthly_dynamic, on='year_month')
comparison_df['dynamic_net_cost_eur'] = comparison_df['dynamic_consumption_cost_eur'] - comparison_df['dynamic_solar_revenue_eur']

# Display comparison table
print(comparison_df[['year_month', 'final_net_cost_eur', 'dynamic_net_cost_eur']].to_string(index=False))

print("\n--- SIMULATION: ANNUAL BOTTOM LINE ---")
print(f"Total Fixed Contract Cost: €{comparison_df['final_net_cost_eur'].sum():.2f}")
print(f"Total Dynamic Contract Cost: €{comparison_df['dynamic_net_cost_eur'].sum():.2f}")
```

## To get the absolute, true-to-life bottom-line total, we need to fold in the fixed annual items:

<li>Liander network connection fees</li>

<li>Fixed supplier delivery fees</li>

<li>Minus the annual Dutch government energy tax credit (teruggave energiebelasting, which is roughly €635.19)</li>

```python
# 1. Ensure year_month is set
energy_minute_df['year_month'] = energy_minute_df['utc_time'].dt.tz_localize(None).dt.to_period('M')

# 2. Group actuals from enhanced master file safely
agg_dict = {
    'energy_consumed_kwh': lambda x: (energy_minute_df.loc[x.index, 'power_consumed_kw'] / 60.0).sum(),
    'energy_produced_kwh': lambda x: (energy_minute_df.loc[x.index, 'power_produced_kw'] / 60.0).sum(),
    'actual_consumption_cost_eur': 'sum',
    'actual_solar_revenue_eur': 'sum'
}
if 'total_tlk_eur' in energy_minute_df.columns:
    agg_dict['total_tlk_eur'] = 'sum'

monthly_financials = energy_minute_df.groupby('year_month').agg(agg_dict).reset_index()

# If total_tlk_eur wasn't in the parquet, default it to 0
if 'total_tlk_eur' not in monthly_financials.columns:
    monthly_financials['total_tlk_eur'] = 0.0

# 3. Group dynamic simulation
df_dynamic['year_month'] = df_dynamic['utc_time'].dt.tz_localize(None).dt.to_period('M')
monthly_dynamic = df_dynamic.groupby('year_month').agg({
    'dynamic_consumption_cost_eur': 'sum',
    'dynamic_solar_revenue_eur': 'sum'
}).reset_index()

# 4. Merge tables
comparison_df = pd.merge(monthly_financials, monthly_dynamic, on='year_month')

# 5. Add fixed annual items allocated monthly
MONTHLY_STANDING_FEES = 35.00
MONTHLY_TAX_CREDIT = 635.19 / 13.0 

comparison_df['fixed_contract_total_eur'] = (
    comparison_df['actual_consumption_cost_eur'] +
    comparison_df['total_tlk_eur'] -
    comparison_df['actual_solar_revenue_eur'] +
    MONTHLY_STANDING_FEES -
    MONTHLY_TAX_CREDIT
)

comparison_df['dynamic_contract_total_eur'] = (
    comparison_df['dynamic_consumption_cost_eur'] -
    comparison_df['dynamic_solar_revenue_eur'] +
    MONTHLY_STANDING_FEES -
    MONTHLY_TAX_CREDIT
)

# 6. Display the final bottom-line comparison
display_cols = ['year_month', 'fixed_contract_total_eur', 'dynamic_contract_total_eur']
print(comparison_df[display_cols].round(2).to_string(index=False))

print("\n--- ANNUAL BOTTOM LINE ---")
print(f"Total Fixed Contract Cost: €{comparison_df['fixed_contract_total_eur'].sum():.2f}")
print(f"Total Dynamic Contract Cost: €{comparison_df['dynamic_contract_total_eur'].sum():.2f}")
```

## The Smart Appliance "What-If" Simulation

The Goal: Testing what happens if you automatically shift heavy appliances (washing machines, dishwashers) to the 11:00 AM – 3:00 PM solar peak window instead of running them in the evening.

Instead of needing historical logs of when your washing machine actually turned on, a smart-home optimization simulation works by synthetically shifting a heavy appliance load (like a 1.5 kWh washing machine cycle taking 1 to 2 hours) from a typical high-cost evening slot into your peak solar window (e.g., between 11:00 AM and 3:00 PM).

By doing this, you calculate two things:

<b>The Baseline Cost</b>: Running the appliance in the evening when you draw expensive grid power (plus evening tariffs/fees).

<b>The "What-If" Cost</b>: Running it during peak solar hours when your panels are producing, meaning the appliance drinks your own solar energy for free instead of forcing that solar power backward onto the grid to incur terugleverkosten (penalty fees).

```python
# 1. Make a copy for our simulation
sim_df = df_dynamic.copy()
sim_df['utc_time'] = pd.to_datetime(sim_df['utc_time'])
sim_df['hour'] = sim_df['utc_time'].dt.hour

# 2. Filter for peak solar vs evening hours
peak_solar_hour = sim_df[sim_df['hour'] == 13]
evening_hour = sim_df[sim_df['hour'] == 20]

# 3. Calculate financial savings from shifting a heavy appliance load
# Shifting from evening grid draw (paying spot price) to mid-day solar self-consumption (avoiding penalty/export waste)
avg_spot_price_evening = evening_hour['spot_price_eur_kwh'].mean()
avg_penalty_day = 0.05  # Estimated average terugleverkosten penalty fee per kWh during peak hours

appliance_kwh = 1.5  # kWh per standard appliance cycle (e.g., washing machine)
days_simulated = sim_df['utc_time'].dt.date.nunique()

estimated_savings = days_simulated * appliance_kwh * (avg_spot_price_evening + avg_penalty_day)

print("--- SMART APPLIANCE 'WHAT-IF' SIMULATION ---")
print(f"Days Analyzed: {days_simulated}")
print(f"Estimated savings from shifting a daily 1.5 kWh appliance cycle to 1:00 PM: €{estimated_savings:.2f}")
```

## The Status Quo Scenario & Dynamic Simulation for 2027

**Scenario 1: Status Quo Bottom-Line Spend (Fixed Contract + Penalties + Fees & Credits)**

$$\text{Total Spend} = \text{Consumption Cost} + \text{Export Penalties (TLK)} - \text{Solar Revenue} + \text{Standing Fees} - \text{Government Tax Credit}$$

*What it means:* You pay for grid consumption, add 2027 export penalties (*terugleverkosten*) for pushing solar back, subtract feed-in revenues, add monthly fixed supplier/network fees, and deduct the Dutch government's annual energy tax credit (*vermindering energiebelasting*)

**Scenario 2: Optimized Dynamic Bottom-Line Spend (Dynamic Contract + Load Shifting + Fees & Credits)**

$$\text{Total Spend} = \text{Dynamic Consumption Cost} - \text{Dynamic Solar Revenue} + \text{Standing Fees} - \text{Government Tax Credit}$$

*What it means:* You pay for hourly dynamic consumption minus optimized solar injection revenue (maximized by shifting heavy loads to cheap or negative spot-price hours), combined with the same fixed standing fees and government tax credits.
