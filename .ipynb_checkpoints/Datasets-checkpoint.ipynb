## 1. Weather Dataset

Weather & Solar Radiation Telemetry (KNMI) for the Schipol weather station (240) as part of MVP.<br>

Why we add weather data:<br>
Solar panels depend on the weather, especially sunshine and clouds. <br>
By combining local weather data with household energy records, we can:<br>
<ul>a. Predict when the house will make extra solar energy (which causes feed-in penalty fees).</ul>
<ul>b. See how outside temperatures change how much electricity the house uses for heating or cooling.</ul>

<i>Sources</i>:<br>
<a href="https://www.daggegevens.knmi.nl/klimatologie/uurgegevens">KNMI</a><br>

## Understanding the weather data

<b>Temperature</b>: (in 0.1 degrees Celsius) at a height of 1.50 m during the observation

<b>Sunshine_hrs</b>: Duration of sunshine (in 0.1 hours) per hour interval, calculated from global radiation (-1 for <0.05 hours)

<b>Global_radiation</b>: (in J/cm²) per hour interval

<b>Cloud_cover</b>: Cloud cover (upper air coverage in eighths), during observation (9=upper air invisible)

```python
# Convert temperature into actual celsius value by dividing by 10

weather_history_df["Temperature"] = weather_history_df["Temperature"] / 10
```

```python
# Convert sunshine hours into actual real value by dividing by 10

weather_history_df["Sunshine_hrs"] = weather_history_df["Sunshine_hrs"] / 10
```

```python
weather_history_df.sample()
```

```python
# Convert into datetime

# Handle KNMI's 1-24 hour format (if hour is 24, shift date forward by 1 day and set hour to 0)
is_hour_24 = weather_history_df['Time'] == 24

# Convert Date int to datetime
base_date = pd.to_datetime(weather_history_df['Date'].astype(str), format='%Y%m%d')

# Adjust dates where hour was 24
adjusted_date = base_date + pd.to_timedelta(is_hour_24.astype(int), unit='days')

adjusted_hour = weather_history_df['Time'].replace(24, 0)

# Create the final unified datetime column - A single timestamp makes exact-match merging much easier down the line
weather_history_df['timestamp'] = adjusted_date + pd.to_timedelta(adjusted_hour, unit='h')
```

```python
# Extract temporal features

weather_history_df['day_of_week'] = weather_history_df['timestamp'].dt.day_name()  # 'Monday', 'Tuesday', etc.
weather_history_df['is_weekend'] = (weather_history_df['timestamp'].dt.dayofweek >= 5)  # True/False

weather_history_df['month'] = weather_history_df['timestamp'].dt.month
weather_history_df['hour'] = weather_history_df['timestamp'].dt.hour

print(weather_history_df[['timestamp', 'day_of_week', 'is_weekend', 'Temperature']].head())
```

## 2.2 Home Assitant
Source: Home Assistant P1 / Solar export - High frequency (1-minute intervals)


<b>1 minute intervals:</b>

<i>power_consumed_kw, power_produced_kw  :</i> Total consumption & production

<i>power_consumed_phase_1_kw, power_produced_phase_1_kw  :</i> Phase 1 consumption & production

<i>solar_output_w, pv_all_power_w, pv1_power_w:</i> Solar specific tracking. Only installed in August 2026

```python
# Load the Home assistant dataset: High-frequency (1-minute intervals)

energy_minute_df = pd.read_csv("power_1min_2025-10-01_2026-10-01.csv")

energy_minute_df.head()
```

```python
energy_minute_df.info()
```

```python
# Convert to datetime

energy_minute_df['utc_time'] = pd.to_datetime(energy_minute_df['utc_time'])
energy_minute_df = energy_minute_df.sort_values('utc_time').reset_index(drop=True)
```

```python
# Handle missing values

energy_minute_df.isnull().sum()
```

```python
# Clean the solar watt columns so that before 11 August 2026, it remains NaN but after it was installed it measures 0.0 
# (this could be because of a reading at night)

# Define the cutoff date (11 August 2026)
cutoff_date = pd.to_datetime('2026-08-11').date()
is_after_install = energy_minute_df['utc_time'].dt.date >= cutoff_date

# Standardize units of measurement and convert watts to kilowatts
solar_watt_cols = ['solar_output_w', 'pv_all_power_w', 'pv1_power_w']

for col in solar_watt_cols:
    # Fill NaNs with 0.0 ONLY for dates on or after August 11, 2026
    energy_minute_df.loc[is_after_install & energy_minute_df[col].isna(), col] = 0.0
    
    # Convert to kW
    energy_minute_df[col] = energy_minute_df[col] / 1000.0
    
    # Rename to _kw
    energy_minute_df = energy_minute_df.rename(columns={col: col.replace('_w', '_kw')})
```

```python
energy_minute_df.isnull().sum()
```

## Why are there missing values:
<i>power_consumed_kw</i> (Home Power Draw): The house is always using electricity (the fridge, router, standby devices), so missing values here are usually just brief telemetry dropouts where Home Assistant or the smart meter missed a single minute recording.

<i>power_produced_kw</i> (Grid Export): You aren't always pushing power back to the grid. If you aren't exporting electricity, the meter doesn't log anything, resulting in a NaN.

```python
# Clean the power_consumed and power_produced columns

# Clean consumption: Forward-fill brief dropouts (up to 5 minutes), then fill any remaining with 0.0
consumption_cols = ['power_consumed_kw', 'power_consumed_phase_1_kw']
for col in consumption_cols:
    energy_minute_df[col] = energy_minute_df[col].ffill(limit=5).fillna(0.0)

# Clean production (grid export): Fill missing values with 0.0 (meaning no export at that minute)
production_cols = ['power_produced_kw', 'power_produced_phase_1_kw']
for col in production_cols:
    energy_minute_df[col] = energy_minute_df[col].fillna(0.0)
```

```python
# Run a final check on null counts to make sure everything is clean

print(energy_minute_df.isnull().sum())
```

## 3.1 Fixed Energy Pricing with Vattenfall and Greenchoice


## Assign the Suppliers

The contract switched from Vattenfall to Greenchoice in February 2026. Apply a filer for the following:

Vattenfall rates: 1 October 2025 - 22 February 2026

Greenchoice rates: 23 February 2026 - 1 October 2026


```python
# 1. Define a simple function to decide the supplier based on the date

def assign_supplier(row_time):
    # If the date is between 1 Oct 2025 and 22 Feb 2026, it's Vattenfall
    if row_time >= pd.Timestamp('2025-10-01', tz='UTC') and row_time < pd.Timestamp('2026-02-23', tz='UTC'):
        return 'Vattenfall'
    else:
        # Otherwise, it's Greenchoice
        return 'Greenchoice'

# 2. Apply this function to every row in your 'utc_time' column
merged_energy_df['supplier'] = merged_energy_df['utc_time'].apply(assign_supplier)

# 3. Check the result
print(merged_energy_df['supplier'].value_counts())
```

## Baseline Reality: Calculate Consumption Costs & Solar Revenues for Vattenfall & Greenchoice

<i>Vattenfall</i><br>
Fixed consumption rate1: 0.25971<br>
Fixed consumption rate2: 0.29383<br>

Solar feed-in compensation: 0.168

<i>Greenchoice</i><br>
Fixed consumption rate1: 0.23443<br>
Fixed consumption rate2: 0.22233<br>

Solar feed-in compensation minus terugleverkosten: 0.013

```python
# Apply fixed contract rates for the actual baseline
# (Using the Vattenfall and Greenchoice blended rates from the contract sheets)

def calculate_actual_costs_split(row):
    # Retrieve consumption for both tariffs per minute (kWh)
    kwh_consumed_t1 = row.get('consumed_tariff_1_kwh', 0.0) / 60.0  # Off-peak / Dal
    kwh_consumed_t2 = row.get('consumed_tariff_2_kwh', 0.0) / 60.0  # Peak / Normaal
    kwh_produced = row.get('power_produced_kw', 0.0) / 60.0
    
    if row['supplier'] == 'Vattenfall':
        rate_t1 = 0.25971  # Off-peak rate (€/kWh)
        rate_t2 = 0.29383  # Peak rate (€/kWh)
        solar_revenue = kwh_produced * 0.16800
        
    else:  # Greenchoice
        rate_t1 = 0.23443  # Off-peak rate (€/kWh)
        rate_t2 = 0.22233  # Peak rate (€/kWh)
        net_solar_rate = 0.15 - 0.13613
        solar_revenue = kwh_produced * net_solar_rate
        
    consumption_cost = (kwh_consumed_t1 * rate_t1) + (kwh_consumed_t2 * rate_t2)
    return pd.Series([consumption_cost, solar_revenue])

# Apply the split calculation across your dataframe
merged_energy_df[['actual_consumption_cost_eur', 'actual_solar_revenue_eur']] = merged_energy_df.apply(calculate_actual_costs_split, axis=1)

print("Exact tariff split calculations completed successfully.")
```

```python
# Apply fixed contract rates for the actual baseline
# (Using your Vattenfall and Greenchoice blended rates from the contract sheets)

def calculate_actual_costs(row):
    # Convert kW to kWh for this minute
    kwh_consumed = row['power_consumed_kw'] / 60.0
    kwh_produced = row['power_produced_kw'] / 60.0
    
    if row['supplier'] == 'Vattenfall':
        # Vattenfall rates (using normaaltarief / daltarief logic or a blended fixed rate)
        # For simplicity I'm using the all-in peak rate (~€0.29383/kWh)
        consumption_cost = kwh_consumed * 0.29383
        # Vattenfall solar feed-in compensation (€0.168/kWh)
        solar_revenue = kwh_produced * 0.16800
        
    else:
        # Greenchoice rates (using enkel/normaal/dal rate, e.g., ~€0.22838/kWh)
        consumption_cost = kwh_consumed * 0.22838
        # Greenchoice solar: (compensation €0.15) minus (terugleverkosten €0,13613) = net ~€0.01387/kWh
        net_solar_rate = 0.15 - 0.13613
        solar_revenue = kwh_produced * net_solar_rate
        
    return pd.Series([consumption_cost, solar_revenue])

# Apply the calculation across your dataframe
merged_energy_df[['actual_consumption_cost_eur', 'actual_solar_revenue_eur']] = merged_energy_df.apply(calculate_actual_costs, axis=1)

print("Actual baseline costs calculated successfully.")
```

## Calculate Terugleverkosten and Fixed Fees

```python
# 1. Group by supplier to see total consumption and solar export for each period
supplier_summary = merged_energy_df.groupby('supplier').agg({
    'energy_consumed_kwh': 'sum',
    'energy_produced_kwh': 'sum',
    'actual_consumption_cost_eur': 'sum',
    'actual_solar_revenue_eur': 'sum'
})

print("--- Summary by Supplier Period ---")
print(supplier_summary)
```

```python
# 2. Calculate Terugleverkosten (TLK) per supplier rules:
# --- Vattenfall (Oct 2025 - Feb 2026): Fixed bracket fee ---
# Looking at the Vattenfall rates, Schaal 4 (750-1000 kWh) was €121.00/year 
# Since this period is roughly 5 months, we can prorate it or apply the flat fee if billed annually.
vattenfall_export = supplier_summary.loc['Vattenfall', 'energy_produced_kwh']
vattenfall_tlk = 121.00 * (5/12) # Prorated for the 5 months

# --- Greenchoice (Feb 2026 - Sep 2026): Per-kWh fee ---
# Greenchoice charge: €0.13613 per kWh
greenchoice_export = supplier_summary.loc['Greenchoice', 'energy_produced_kwh']
greenchoice_tlk = greenchoice_export * 0.13613

total_tlk = vattenfall_tlk + greenchoice_tlk
print(f"\nCalculated Vattenfall TLK portion: €{vattenfall_tlk:.2f}")
print(f"Calculated Greenchoice TLK portion: €{greenchoice_tlk:.2f}")
print(f"Total Actual Terugleverkosten: €{total_tlk:.2f}")
```

```python
# 3. Factor in fixed annual baseline items (Liander network fees + fixed delivery - tax credit)
# From the Vattenfall sheet: Liander = €455.74, Fixed delivery = €107.88, Tax credit = -€635.19

net_fixed_annual = 455.74 + 107.88 - 635.19  # Total net fixed cost/credit for the year

# 4. Calculate Final Actual Baseline Total
total_actual_consumption = supplier_summary['actual_consumption_cost_eur'].sum()
total_actual_solar_rev = supplier_summary['actual_solar_revenue_eur'].sum()

final_actual_cost = total_actual_consumption + total_tlk + net_fixed_annual - total_actual_solar_rev

print(f"\n========================================")
print(f"FINAL ACTUAL HISTORICAL ENERGY SPEND: €{final_actual_cost:.2f}")
print(f"========================================")
```

## 3.2 Dynamic Energy Pricing for the Netherlands

To pull live or historical dynamic energy prices for the Netherlands into your pipeline, the gold standard source used by data engineers across Europe is the ENTSO-E Transparency Platform API (the official clearinghouse for European wholesale electricity data).

<i>Source</i>: <a href="transparency.entsoe.eu">ENTSO-E Transparency Platform</a>

<i>Fields</i>: spot_price_eur_kwh, is_negative_pricing

<b>Negative Pricing</b>: Wholesale spot prices drop below zero when renewable generation (solar/wind) significantly exceeds grid demand. During these intervals, grid injection incurs a cost rather than earning revenue.

```python
# Load the .env file with the keys
load_dotenv()
client = Client(api_key=os.environ.get("ENTSOE_API_KEY"))
```

```python
# 1. Define your full-year timeframe
start_date = pd.Timestamp("2025-10-01", tz="Europe/Brussels")
end_date = pd.Timestamp("2026-11-01", tz="Europe/Brussels") # End is 1 November because freq='Ms' (month start)

# 2. Generate monthly start dates to chunk the requests safely
month_ranges = pd.date_range(start=start_date, end=end_date, freq='MS')

all_market_dfs = []

# 3. Loop through month by month
for i in range(len(month_ranges) - 1):
    chunk_start = month_ranges[i]
    chunk_end = month_ranges[i+1]
    
    print(f"Fetching market prices for {chunk_start.strftime('%Y-%m')}...")
    
    try:
        # Pull the month chunk
        df_chunk = client.prices.day_ahead(chunk_start, chunk_end, country="NL")
        
        # Clean and convert each chunk immediately
        df_temp = df_chunk.reset_index()
        time_col = 'timestamp' if 'timestamp' in df_temp.columns else 'index'
        price_col = 'value' if 'value' in df_temp.columns else df_temp.columns[1]
        
        df_temp = df_temp[[time_col, price_col]].copy()
        df_temp.columns = ["timestamp", "spot_price_eur_kwh"]
        
        all_market_dfs.append(df_temp)
    except Exception as e:
        print(f"Failed to fetch {chunk_start.strftime('%Y-%m')}: {e}")

# 4. Concatenate all months into one master dataframe
df_dynamic_market = pd.concat(all_market_dfs, ignore_index=True)

# 5. Final transformations (Unit conversion & negative pricing flag)
df_dynamic_market["spot_price_eur_kwh"] = df_dynamic_market["spot_price_eur_kwh"] / 1000.0
df_dynamic_market["is_negative_pricing"] = df_dynamic_market["spot_price_eur_kwh"] < 0

print("\n--- Full Year Loaded Successfully! ---")
print(f"Total hourly/15-min price rows: {len(df_dynamic_market)}")
print(df_dynamic_market.head())
```

## Merging the energy_minute_df with df_dynamic_market
Since the smart meter data is recorded every minute, and the market prices update every 15 minutes, the best way to combine them without losing resolution is a backward merge / asof merge or reindexing so that every 1-minute row inherits the active 15-minute market price of its current window.

```python
# 1. Clean and ensure correct datetime formatting & sorting
energy_minute_df['utc_time'] = pd.to_datetime(energy_minute_df['utc_time'])
df_dynamic_market['timestamp'] = pd.to_datetime(df_dynamic_market['timestamp'])

# Ensure UTC alignment
energy_minute_df['utc_time'] = energy_minute_df['utc_time'].dt.tz_convert('UTC')
df_dynamic_market['timestamp'] = df_dynamic_market['timestamp'].dt.tz_convert('UTC')

# CRITICAL: Sort both ascendingly so 'backward' lookups look forward in real time
energy_minute_df = energy_minute_df.sort_values('utc_time').reset_index(drop=True)
df_dynamic_market = df_dynamic_market.sort_values('timestamp').reset_index(drop=True)

# 2. Perform the asof merge
merged_energy_df = pd.merge_asof(
    energy_minute_df,
    df_dynamic_market,
    left_on='utc_time',
    right_on='timestamp',
    direction='backward'
)

# 3. Quick verification print to check matching dates
print(merged_energy_df[['utc_time', 'timestamp', 'spot_price_eur_kwh']].head(10))
```

## Calculate Consumption Costs & Solar Revenues

For a dynamic energy contract in the Netherlands, your total cost per minute is typically:

Consumption Cost: Energy consumed ($\text{kWh}$) $\times$ (Spot Price + Supplier Markup + Government Energy Tax & VAT).

Injection Revenue: Energy injected back into the grid ($\text{kWh}$) $\times$ Spot Price (though subject to terugleverkosten).

```python
# Define the contract parameters (adjust these to your actual supplier terms)
SUPPLIER_MARKUP_EUR_KWH = 0.025  # Example Inkoopvergoeding / opslag per kWh
ENERGY_TAX_EUR_KWH = 0.1316      # Example Dutch energy tax + ODE per kWh (excluding VAT)

# 1. Consumption financial calculation
# Total price per kWh consumed = Spot Price + Supplier Markup + Government Tax
merged_energy_df['total_consumer_price_kwh'] = (
    merged_energy_df['spot_price_eur_kwh'] + 
    SUPPLIER_MARKUP_EUR_KWH + 
    ENERGY_TAX_EUR_KWH
)

merged_energy_df['consumption_cost_eur'] = (
    merged_energy_df['energy_consumed_kwh'] * 
    merged_energy_df['total_consumer_price_kwh']
)

# 2. Solar export financial calculation
# Revenue from feeding power back into the grid at the spot price
merged_energy_df['solar_revenue_eur'] = (
    merged_energy_df['energy_produced_kwh'] * 
    merged_energy_df['spot_price_eur_kwh']
)
```

```python
merged_energy_df.tail()
```

```python
# Monthly summary grouped by Year-month

# 1. Create a year-month period column (e.g., '2025-10', '2025-11')
merged_energy_df['year_month'] = merged_energy_df['utc_time'].dt.to_period('M')

# 2. Group by month and aggregate (summing up kWh and Euros)
monthly_financials = merged_energy_df.groupby('year_month', as_index=False).agg({
    'energy_consumed_kwh': 'sum',
    'energy_produced_kwh': 'sum',
    'consumption_cost_eur': 'sum',
    'solar_revenue_eur': 'sum'
})

# Calculate net monthly cost (consumption costs minus solar revenues)
monthly_financials['net_cost_eur'] = (
    monthly_financials['consumption_cost_eur'] - 
    monthly_financials['solar_revenue_eur']
)

print("--- Monthly Energy & Financial Summary ---")
print(monthly_financials.to_string(index=False))
```
